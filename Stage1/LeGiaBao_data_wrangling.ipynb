```python
import pandas as pd
import numpy as np
```

# Data Wrangling

## Objective

The purpose of this section is to transform the cleaned dataset from the previous step into a structured, consistent, and validated dataset for Exploratory Data Analysis (EDA).

The data wrangling process consists of six main steps:

1. Discovery
2. Structuring
3. Cleaning
4. Enriching
5. Validating
6. Publishing

## 1. Discovery

The first step is to explore the dataset received from the previous missing-value processing step.

The objectives are to:
- Check the number of rows and columns.
- Inspect the column names and data types.
- Review the first records.
- Check duplicated rows
- Examine the basic statistics of numerical variables.
- Review the values contained in categorical variables.

```python
df = pd.read_csv("../data/LeDangKhoa_cleaned_missing.csv")

print("Dataset shape:", df.shape)
print("\nColumn names:")
print(df.columns.tolist())

print("\nData types:")
print(df.dtypes)

print("\nFirst 5 rows:")
display(df.head())
```

```python
# Check basic statistics
display(df.describe(include="all").T)
```

```python
# Check duplicated records
print("Duplicate rows:", df.duplicated().sum())
```

```python
# Basic statistics for numerical columns

display(
    df[["Quantity", "Price"]].describe()
)
```

```python
# Check unique values in categorical columns

for col in ["Product Name", "Store Location", "Source"]:
    print(f"\n{col}:")
    print("Number of unique values:", df[col].nunique())
    print(df[col].value_counts().head(15))
```

## 2. Structuring

The dataset already contains a consistent set of seven columns after the previous processing step.

In this step, the structure is standardized by:
- Removing unnecessary spaces from text values.
- Converting `Transaction Date` into datetime format.
- Ensuring `Quantity` and `Price` are numeric.
- Keeping the original column names and dataset structure so that the following analysis stages can use the same variables consistently.

```python
# Remove unnecessary leading and trailing spaces
# from text columns

text_columns = [
    "Transaction ID",
    "Product Name",
    "Store Location",
    "Source"
]

for col in text_columns:
    df[col] = df[col].astype(str).str.strip()


# Convert Transaction Date to datetime

df["Transaction Date"] = pd.to_datetime(
    df["Transaction Date"],
    errors="coerce"
)


# Ensure numerical columns are numeric

df["Quantity"] = pd.to_numeric(
    df["Quantity"],
    errors="coerce"
)

df["Price"] = pd.to_numeric(
    df["Price"],
    errors="coerce"
)


print("Data types after structuring:")
print(df.dtypes)
```

```python
# Check the structured dataset

print("Shape:", df.shape)

print("\nFirst 5 rows:")
display(df.head())
```

## 3. Cleaning

The missing values have already been handled in the previous step.

Therefore, this step focuses on checking and cleaning duplicated or logically invalid records.

The cleaning process includes:
- Removing duplicated rows.
- Checking duplicated Transaction IDs.
- Checking invalid values in `Quantity` and `Price`.
- Checking invalid transaction dates.

If no invalid records are found, the dataset is kept unchanged.

```python
# Remove duplicated rows

rows_before = len(df)

df = df.drop_duplicates().reset_index(drop=True)

rows_after = len(df)

print("Rows before removing duplicates:", rows_before)
print("Rows after removing duplicates:", rows_after)
print("Duplicated rows removed:", rows_before - rows_after)
```

```python
# Check duplicated Transaction IDs

duplicate_ids = df["Transaction ID"].duplicated().sum()

print("Duplicated Transaction IDs:", duplicate_ids)
```

```python
# Check logically invalid numerical values

invalid_quantity = (df["Quantity"] <= 0).sum()
invalid_price = (df["Price"] <= 0).sum()

print("Invalid Quantity values:", invalid_quantity)
print("Invalid Price values:", invalid_price)
```

```python
# Check invalid transaction dates

invalid_dates = df["Transaction Date"].isna().sum()

print("Invalid Transaction Date values:", invalid_dates)
```

## 4. Enriching

The original dataset contains `Quantity` and `Price`, which can be combined to calculate the sales amount of each transaction.

A new column `Sales Amount` is created using:

**Sales Amount = Quantity × Price**

This derived variable can be used in later analysis to examine sales performance and revenue.

```python
# Calculate Sales Amount for each transaction

df["Sales Amount"] = (
    df["Quantity"] * df["Price"]
)

display(
    df[
        [
            "Quantity",
            "Price",
            "Sales Amount"
        ]
    ].head()
)
```

```python
# Check the calculated Sales Amount

sales_check = (
    df["Sales Amount"]
    == df["Quantity"] * df["Price"]
)

print(
    "Incorrect Sales Amount calculations:",
    (~sales_check).sum()
)
```

## 5. Validating

The validation step checks whether the dataset satisfies the expected data quality requirements after wrangling.

The following conditions are checked:
- No missing values remain.
- `Quantity` contains valid positive values.
- `Price` contains valid positive values.
- `Sales Amount` is correctly calculated.
- The values of `Store Location` and `Source` are reviewed for consistency.

```python
# Check missing values

print("Missing values in each column:")
display(df.isna().sum())
```

```python
# Validate numerical values

print(
    "Invalid Quantity:",
    (df["Quantity"] <= 0).sum()
)

print(
    "Invalid Price:",
    (df["Price"] <= 0).sum()
)
```

```python
# Validate Sales Amount

sales_check = (
    df["Sales Amount"]
    == df["Quantity"] * df["Price"]
)

print(
    "Incorrect Sales Amount calculations:",
    (~sales_check).sum()
)
```

```python
# Review categorical values

print("Store Location values:")
print(df["Store Location"].value_counts())

print("\nSource values:")
print(df["Source"].value_counts())
```

## 6. Publishing

After completing the data wrangling and validation process, the final dataset is exported as a CSV file.

The published dataset will be used as the input for the following EDA stages.


```python
# Export the final wrangled and validated dataset

output_file = "data/LeGiaBao_wrangled_validated.csv"

df.to_csv(
    output_file,
    index=False
)

print("Final dataset exported successfully.")
print("File:", output_file)
print("Shape:", df.shape)
```
